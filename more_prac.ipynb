```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Zara", "Bilal", "Hina"],
    "RollNo": [1, 2, 3, 4, 5, 6],
    "Math": [85, 92, 78, 88, 95, 70],
    "Science": [80, 89, 75, 90, 91, 65],
    "English": [88, 94, 80, 85, 89, 72]
}
df = pd.DataFrame(data)
print(df)
print(df.describe())
print("Top Math Scorer:", df.loc[df['Math'].idxmax(), 'Name'])
```

```python
df.sort_values("Math" , ascending=False) #q k pandas by-default ascending order mein arrange kr deta hai
(df.sort_values(["Math", "Science"], ascending=[False, True]))
```
